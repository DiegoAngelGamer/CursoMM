## Ejercicio 2.12

Hacemos la función para representar triángulos o trapecios según el número 

```python
def triangulo(dom):
    dom.sort()
    fig, ax = plt.subplots()

    ax.plot([0, dom[0]], [0, 0], c = 'dodgerblue')
    ax.plot([dom[-1], dom[-1] + 0.5*(dom[-1] - dom[0])], [0, 0])

    for i in range(len(dom) - 1):
        if i == 0:
            ax.plot([dom[i], dom[i + 1]], [0, 1], c = 'dodgerblue')
        elif i == len(dom) - 2:
            ax.plot([dom[i], dom[i + 1]], [1, 0], c = 'dodgerblue')
        else:
            ax.plot([dom[i], dom[i + 1]], [1, 1], c = 'dodgerblue')

    for i in range(1, len(dom) - 1):
        ax.plot([dom[i], dom[i]], [0, 1], '--', c = 'blue')
        ax.plot([dom[i], dom[i]], [0, 1], 'o', c = 'dodgerblue')
    ax.set_xticks(dom)
```

```python
triangulo([3,4,5,6])
```
